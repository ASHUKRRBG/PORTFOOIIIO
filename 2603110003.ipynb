# ** program -1**
The Setup: n = 1 initializes the variable at 1.
The Guard: while n < 10: checks if n is strictly less than 10. If it is, the loop is allowed to run.

The Action: Inside the loop, n = n * 2 doubles the value, and print(n) outputs that new value to the screen.

The Sequence:

Pass 1: 1 becomes 2 (Prints 2)

Pass 2: 2 becomes 4 (Prints 4)

Pass 3: 4 becomes 8 (Prints 8)

Pass 4: 8 becomes 16 (Prints 16)

```python
n = 1
while n < 10:
    n = n * 2
    print(n)
print("done", n)
```

 **Program -2 **The Setup: count = 0 creates a variable and sets its starting value to 0.

The Loop: for i in range(3): tells the program to run the indented code block exactly 3 times.

The Action: count = count + 2 adds 2 to the running total during each of those 3 passes (0 → 2, 2 → 4, 4 → 6).

The Finish: print(count) runs after the loop is completely finished, outputting the final result of 6 to the screen.

```python
count = 0
for i in range(3):
    count = count + 2
print(count)
```

** Program-4 **Standard · The Self-Sorting Bench
everyone reaches here
Put Job 09's three-bin ladder inside the loop: each generated diameter is announced with its bin (ACCEPT / REWORK / SCRAP), three counters tally, and a shift report closes.
Then set the loop to 100. Same code, ten times the work, zero extra effort — write that sentence in your record.

```python
import random

# 1. Start with empty bins
accept = 0
rework = 0
scrap = 0

# 2. Make 100 parts
for i in range(100):

    # Generate a random size and round it to 2 decimal places
    diameter = round(random.uniform(24.90, 25.10), 2)

    # 3. Sort the part into the right bin
    if diameter < 24.95:
        scrap = scrap + 1
        print("Size:", diameter, "--> SCRAP")

    elif diameter > 25.05:
        rework = rework + 1
        print("Size:", diameter, "--> REWORK")

    else:
        accept = accept + 1
        print("Size:", diameter, "--> ACCEPT")

# 4. Print the final totals at the very end
print("--- SHIFT REPORT ---")
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```
